```python
print("WELCOME TO THE GUESSING NUMBER GAME")     #original  level 4  character add akumpol error kanikan padila
print("guessing the number between 1-100")
import random

secret_number=random.randint(1,100)
for i in range(0,8):
    try:              #.........
      user_input=int(input("enter the number:"))
      if user_input<1 or user_input>100:
          print("invalid number ! please enter the number between 1-100")
      elif user_input < secret_number:
          print("your number is lesser")
      elif user_input > secret_number:
         print("your number is higher")
      else:
          print("your guess is correct")
          print(f"congrats you have guessed the number {secret_number} in {i} attempts:")
          break
    except ValueError:    #............
        print("invalid input,enter numbers only!")
else:
    print("limit reached,chance over better luck next time!")


```
